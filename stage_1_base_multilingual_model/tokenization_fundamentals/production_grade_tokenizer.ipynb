```python
from transformers import AutoTokenizer

tokenizer = AutoTokenizer.from_pretrained(
    "google-bert/bert-base-uncased"
)
```

```python
print(tokenizer)
```

```python
print("Vocabulary size:", tokenizer.vocab_size)               # number of tokens
print("Model max length:", tokenizer.model_max_length)        # the number of tokens in one sequence.
                                                              # A seq is one tokenized piece of text that we give to BERT as one example
```

```python
tokenizer.tokenize('The patient has patellifemoral chondromalacia')
```

```python
## "##" means that its a continuation of a word, so the 3 words are part of the original word and not whitespace separated tokens
```

```python
text = "The patient has patellofemoral chondromalacia."

tokens = tokenizer.tokenize(text)                      # 'tokenize' will tokenize, i.e. split into tokens

token_ids = tokenizer.convert_tokens_to_ids(tokens)    # will assign numerical ids

for token, token_id in zip(tokens, token_ids):
    print(f"{token:12}:{token_id}")
```

```python
for i in range(15, 20):

    text = dataset["train"][i]["text"]

    tokens = tokenizer.tokenize(text)

    print("=" * 80)
    print("RAW:")
    print(text)

    print("\nTOKENIZED:")
    print(tokens)

    print("\nNumber of tokens:", len(tokens))
```
